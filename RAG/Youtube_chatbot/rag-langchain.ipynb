```python
from youtube_transcript_api import YouTubeTranscriptApi, TranscriptsDisabled
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_openai import OpenAIEmbeddings, ChatOpenAI
from langchain_community.vectorstores import FAISS
from langchain_core.prompts import PromptTemplate
```

```python
video1_id = "o126p1QN_RI"
try:
    ytt_api = YouTubeTranscriptApi()
    transcript_list = ytt_api.fetch(video1_id)

    transcript = " ".join(snippet.text for snippet in transcript_list)
    print(transcript[:500])  # Print the first 500 characters of the transcript
except TranscriptsDisabled:
    print("Transcripts are disabled for this video.")
```

```python
transcript_list
```

step 1b Text Splitting

```python
splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200)
chunks = splitter.create_documents([transcript])
```

```python
len(chunks)
```

```python
chunks[100]
```

step 1c Embedding & Vector Store

```python
embeddings = OpenAIEmbeddings(model="text-embedding-3-small")
vectorstore = FAISS.from_documents(chunks, embeddings)
```

```python
vectorstore.index_to_docstore_id
```

Step 2 - Retrieval

```python
retriever = vectorstore.as_retriever(search_type="similarity", search_kwargs={"k": 5})
```

```python
retriever.invoke("what is RAG?")
```

Step 3 - Augmentation

```python
llm = ChatOpenAI(model="gpt-5.1", temperature=0.9)
```

```python
prompt = PromptTemplate(
    template="""
    You are a helpful assistant.
    Answer Only from the provided transcript context.
    If the context is insufficent, just say you don't know.
    
    {context}
    Question: {question}
    """,
    input_variables=["question" , 'context']
   )
```

```python
question = "is the topic of data ingestion discussed in this video? if yes then what was discussed"
retrieved_docs = retriever.invoke(question)
```

```python
retrieved_docs
```

```python
context_text = "\n\n".join([doc.page_content for doc in retrieved_docs])
```

```python
final_prompt = prompt.invoke({"context": context_text, "question": question})
```

```python
final_prompt
```

Step 4 - Generation

```python
answer = llm.invoke(final_prompt)
print(answer.content)
```

Building Chain

```python
from langchain_core.runnables import RunnableParallel, RunnablePassthrough,  RunnableLambda
```

```python
def format_docs(retrieved_docs):
    context_text = "\n\n".join(doc.page_content for doc in retrieved_docs)
    return context_text
```

```python
parallel_chain = RunnableParallel({
    'context': retriever | RunnableLambda(format_docs),
    'question': RunnablePassthrough()
})
```

```python
parallel_chain.invoke('what is RAG?')
```

```python
from langchain_core.output_parsers import StrOutputParser
parser = StrOutputParser()
```

```python
main_chain = parallel_chain | prompt | llm | parser
```

```python
main_chain.invoke('can you summarize the video?')
```
